```python
# Pandas
import pandas as pd
print("Biblioteka załadowana!")
```

```python
# Dane + BMI
pacjenci = [
    {"imie": "Jan", "waga": 85, "wzrost": 1.80},
    {"imie": "Anna", "waga": 60, "wzrost": 1.65},
    {"imie": "Marek", "waga": 110, "wzrost": 1.75}
]
df = pd.DataFrame(pacjenci)
df['BMI'] = df['waga'] / (df['wzrost'] ** 2)
```

```python
# Kategoria
# Definiujemy progi (bins) i nazwy kategorii (labels)
progi = [0, 18.5, 25, 30, 100]
nazwy = ['Niedowaga', 'Norma', 'Nadwaga', 'Otyłość']

# Nowa kolumna 'Kategoria'
df['Kategoria'] = pd.cut(df['BMI'], bins=progi, labels=nazwy)
```

```python
# Sprawdzam podstawowe statystyki (średnia, odchylenie, min/max)
df.describe()
```
