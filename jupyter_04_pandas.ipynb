Rozdział 18. Łączenie zbiorów danych - concat i append

```python
# Funkcja tworząca ramkę danych
def make_df(cols, ind):
    """Funkcja, która szybko tworzy ramkę danych"""
    data = {c: [str(c) + str(i) for i in ind]
            for c in cols}
    return pd.DataFrame(data, ind)
```

```python
# Przykładowa ramka
make_df('ABC', range(3))
```

```python
# Klasa, która pozwoli wyświetlać obok siebie kilka ramek danych
class display(object):
    """Funkcja, wyswietlająca reprezentację HTML wielu obiektów."""
    template = """<div style="float: left; padding: 10px;">
    <p style='font-family:"Courier New", Courier, monospace'>{0}{1}
    """
    def __init__(self, *args):
        self.args = args
    
    def _repr_html_(self):
        return '\n'.join(self.template.format(a, eval(a)._repr_html_())
                         for a in self.args)
    
    def __repr__(self):
        return '\n\n'.join(a + '\n' + repr(eval(a))
                           for a in self.args)
    
```

```python
df1 = make_df('AB', [1, 2])
df2 = make_df('AB', [3, 4])
pd.concat([df1, df2])
```

```python
display('df1', 'df2', 'pd.concat([df1, df2])')
```

```python
df3 = make_df('AB', [0, 1])
df4 = make_df('CD', [0, 1])
display('df3', 'df4', "pd.concat([df3, df4], axis='columns')")
```

```python
x = make_df('AB', [0, 1])
y = make_df('AB', [2, 3])
y.index = x.index # dopasowanie indeksów
display('x', 'y', 'pd.concat([x, y])')
```

```python
try:
    pd.concat([x, y], verify_integrity=True)
except ValueError as e:
    print("ValueError", e)
```

```python
display('x', 'y', 'pd.concat([x, y], ignore_index=True)')
```

```python
display('x', 'y', "pd.concat([x, y], keys=['x', 'y'])")
```

```python
# łaczenie kolumn, gdzie tylko kilka jest wspólnych
df5 = make_df('ABC', [1, 2])
df6 = make_df('BCD', [3, 4])
display('df5', 'df6', 'pd.concat([df5, df6])') # domyślnie argument join='outer'
```

```python
# join='inner'
display('df5', 'df6', "pd.concat([df5, df6], join='inner')")
```

```python
# reindex przed złączeniem
pd.concat([df5, df6.reindex(df5.columns, axis=1)])
```
