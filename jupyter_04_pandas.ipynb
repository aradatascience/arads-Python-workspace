```python
A = pd.Series([2, 4, 6], index=[0, 1, 2])
B = pd.Series([1, 3, 5], index=[1, 2, 3])
A + B
```

```python
A.add(B, fill_value=0)
```

```python
A = pd.DataFrame(rng.integers(0, 20, (2, 2)),
                 columns=['a', 'b'])
A
```

```python
B = pd.DataFrame(rng.integers(1, 10, (3, 3)),
                 columns=['b', 'a', 'c'])
B
```

```python
A + B 
```

```python
A.add(B, fill_value=A.values.mean())
```

```python
A = rng.integers(10, size=(3, 4))
A
```

```python
A - A[0]
```

```python
df = pd.DataFrame(A, columns=['Q', 'R', 'S', 'T'])
df - df.iloc[0]
```

```python
df.subtract(df['R'], axis=0)
```

```python
halfrow = df.iloc[0, ::2]
halfrow
```

```python
df - halfrow
```

Rozdział 17. Indeksowanie hierarchiczne

```python
# indeksowanie wielokrotne inaczej - multiindexing
# typ MultiIndex
```

```python
# Zły sposób na wielokrotne indeksowanie series - KROTKI
index = [('California', 2010), ('California', 2020),
         ('New York', 2010), ('New York', 2020),
         ('Texas', 2010), ('Texas', 2020)]
population = [324242, 2323232, 242424, 23232432, 224242, 242424]
pop = pd.Series(population, index=index)
pop
```

```python
pop[('California', 2020):('Texas', 2010)]
```

```python
# ale juz wybranie wszystkiego z 2010 jest mozolne
pop[[i for i in pop.index if i[1] == 2010]]
```

```python
# Lepszy sposób - MultiIndex z Pandas
index = pd.MultiIndex.from_tuples(index)
```

```python
pop = pop.reindex(index)
pop
```

```python
# mozna zastosowac slicing tutaj
pop[:, 2020]
```
