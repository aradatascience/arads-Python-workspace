Rozdział 14. Indeksowanie i wybieranie

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0],
                 index=['a', 'b', 'c', 'd'])
data
```

```python
data['b']
```

```python
'a' in data
```

```python
data.keys()
```

```python
list(data.items())
```

```python
data['e'] = 1.25
data
```

```python
# slicing z uzyciem jawnego indeksu
data['a':'c']
```

```python
# slicing z uzyciem niejawnego indeksu liczbowego
data[0:2]
```

```python
# maskowanie - filtrowanie danych za pomoca warunkow logicznych (tworzy 'maske' - tablice z True i False)
# nałoenie maski (filtru)
data[(data > 0.3) & (data < 0.8)]
```

```python
# fancy indexing (zaawansowane indeksowanie - zamiast filtrowac indeksami od-do, podajemy konkretne indeksy w swojej kolejnosci)
data[['a', 'e']]
```

Indeksatory: loc i iloc

```python
# to atrybuty aby bylo wiadomo czy uzywamy jawnego czy niejawnego indeksu
# loc - pozwala indeksowac i wycinać fragmenty obiektu na podstawie JAWNEGO indeksu

data = pd.Series(['a', 'b', 'c'], index=[1, 3, 5])
data
```

```python
# uzyty niejawny indeks
data[1:3]
```

```python
data.loc[1]
```

```python
# uzyty jawny indeks
data.loc[1:3]
```
