```python
# Ustrukturyzowane tablice Numpy

import numpy as np

# przykład osobnych 3 tablic
name = ['Alice', 'Bob', 'Cathy', 'Doug']
age = [25, 45, 37, 19]
weight = [55.0, 85.5, 68.0, 61.5]
```

```python
x = np.zeros(4, dtype=int)
x
```

```python
# wykorzystanie złozonego typu danych w ustrukturyzowanej tablicy
# typ danych wprowadzany metodą słownikową!
data = np.zeros(4, dtype={'names':('name', 'age', 'weight'),
                          'formats':('U10', 'i4', 'f8')})
print(data.dtype)
```

```python
data['name'] = name
data['age'] = age
data['weight'] = weight
print(data)
```

```python
data['name']
```

```python
data[0]
```

```python
data[-1]['name']
```

```python
data[data['age'] < 30]['name']
```

```python
# typ danych metodą złozoną
np.dtype([('name', 'S10'), ('age', 'i4'), ('weight', 'f8')])
```

```python
# tablice rekordów - wygodniejsza składnia, ale wolniej odczytują
data_rec = data.view(np.recarray)
data_rec.age
```
