```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Tworzymy szybkie dane tekstowe
data = {
    'Grupa': ['Kontrolna', 'Kontrolna', 'Badawcza', 'Badawcza'],
    'Poziom_cukru': [95, 102, 115, 121]
}
df_test = pd.DataFrame(data)
```

```python
# Rysujemy wykres
sns.set_theme(style="whitegrid")
sns.barplot(x='Grupa', y='Poziom_cukru', data=df_test)
plt.title("Test wizualizacji w środowisku data_science")
plt.show()
```

```python
# Dodanie nowej kolumny na podstawie obliczeń
df_test['Poziom_mmol'] = df_test['Poziom_cukru'] / 18
df_test
```
